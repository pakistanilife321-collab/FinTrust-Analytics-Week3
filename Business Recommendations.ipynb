# Part F — Management Recommendations

The following 5 recommendations are developed directly from the validated gaps, SQL/Python analytics, and Power BI Star Schema modeling requirements.

---

### Recommendation 1: Implement Dynamic RFM Behavioral Customer Segmentation

* **Finding:** Customer segmentation based solely on static demographics fails to capture true customer behavior, recency, or churn risk.
* **Evidence:** Week 2 relied on static demographic fields; Week 3 requires SQL window functions ($NTILE$, $DENSE\_RANK$) and Python RFM clustering.
* **Business Implication:** Marketing and retention budgets are inefficiently spent on dormant or low-value accounts while high-value, decaying accounts are missed.
* **Recommended Action:**
1. Use **SQL** ($NTILE$) and **Python (Pandas/NumPy)** to compute dynamic Recency, Frequency, and Monetary (RFM) scores for all active accounts.
2. Implement $k$-means or quartile-based RFM clustering to group users into actionable tiers (e.g., *Champions*, *At-Risk*, *Hibernating*).
3. Integrate these dynamic tiers into **Power BI** for automated marketing trigger campaigns.



---

### Recommendation 2: Deploy Multi-Dimensional Fraud & Risk Review Monitoring

* **Finding:** Evaluating gross transaction volumes without segmenting by risk status conceals high failure and fraud rates in digital channels.
* **Evidence:** Aggregated channel analysis in Week 2 missed risk status trends; Week 3 requires SQL CTEs joining transaction status with `Risk_Review_Flag` across channels.
* **Business Implication:** High exposure to fraudulent transactions leads to financial losses, higher chargeback fees, and reputational damage.
* **Recommended Action:**
1. Write **SQL CTEs** to compute real-time risk ratios ($\frac{\text{Flagged Transactions}}{\text{Total Transactions}}$) per channel and cohort.
2. Visualize risk distributions using **Seaborn/Matplotlib** heatmaps in **Jupyter Notebook/Google Colab** to detect anomalies.
3. Establish automated alerts in **Power BI** when a channel's risk ratio exceeds pre-set risk-tolerance thresholds.



---

### Recommendation 3: Re-Architect Power BI Dashboard into a Star Schema with Explicit DAX Measures

* **Finding:** Legacy Power BI reports lacked cross-filtering logic, multi-level drill-downs, and dynamic time-intelligence capabilities.
* **Evidence:** Week 2 visualizations were static; Week 3 requires Star Schema data modeling and explicit DAX measures for MTD/YTD growth and channel efficiencies.
* **Business Implication:** Executives lack visibility into true performance drivers, hindering timely strategic and operational decision-making.
* **Recommended Action:**
1. Re-structure the Power BI data model into a normalized **Star Schema** (Fact table for transactions surrounded by Dimension tables for Date, Customer, Channel, and Risk).
2. Write explicit **DAX** measures for $MTD\_Sales$, $YTD\_Sales$, $YOY\_Growth\%$, and $Risk\_Adjusted\_Margin$.
3. Enable bidirectional cross-filtering and multi-level drill-downs from high-level summaries down to individual transaction cohorts.



---

### Recommendation 4: Optimize Channel Operations Using Risk-Adjusted Efficiency Metrics

* **Finding:** Evaluating channels purely on gross processing volume ignores the operational cost of managing risk flags and failed transactions.
* **Evidence:** Week 2 channel evaluation was uncalibrated; Week 3 introduces explicit channel efficiency DAX measures integrated with risk review flags.
* **Business Implication:** Over-investing in high-gross, high-risk channels degrades overall net profitability.
* **Recommended Action:**
1. Calculate net settled revenue per channel using **Pandas** and **SQL** by deducting flagged/reversed amounts from gross volume.
2. Establish a *Channel Risk-Efficiency Index* in **Power BI** using DAX.
3. Reallocate digital infrastructure investments toward channels that yield the highest net risk-adjusted return rather than raw gross volume.



---

### Recommendation 5: Establish Automated Cohort-Based Early Warning Systems

* **Finding:** Risk review flags concentrate within specific customer registration cohorts rather than occurring uniformly.
* **Evidence:** Multi-dimensional cohort analysis joining transaction status with synthetic risk flags reveals localized risk spikes in newly onboarded cohorts.
* **Business Implication:** Delayed identification of vulnerable acquisition cohorts allows fraudulent patterns to persist unchecked.
* **Recommended Action:**
1. Build automated **SQL** cohort tracking scripts that measure 30-, 60-, and 90-day risk review flag rates by acquisition channel and date.
2. Plot cohort survival and risk curves in **Python (Matplotlib/Seaborn)** within **Jupyter Notebooks/Google Colab**.
3. Update digital onboarding KYC (Know Your Customer) workflows dynamically for cohorts flagged as high-risk.
